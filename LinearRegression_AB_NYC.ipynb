```python
#Import Libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
#Task 1.1 Loading into Pandas dataframe
df=pd.read_csv("/content/AB_NYC_2019.csv")
df.info()
```

```python
#Task 1.2 Identifying numerical and categorical columns
numerical = df.select_dtypes(include="number").columns
categorical = df.select_dtypes(exclude="number").columns

print("Numerical variables:")
print(numerical)

print("Categorical variables:")
print(categorical)
```

```python
#Task 1.3 identifying columns with null values
df.isnull().sum()
```

```python
#Task 1.3 Calculating percentages of null values for the above columns
df.isnull().sum()/len(df)*100
```

```python
#Task 2.1 dropping the last_review column entirely
df.drop(columns="last_review",inplace=True)
```

```python
#Task 2.2 filling null reviews_per_month with logically suitable value 0
df["reviews_per_month"] = df["reviews_per_month"].fillna(0)
```

```python
#Task 2.3 dropping irrelevant identifier columns for linear regression
df.drop(columns=["id", "host_id", "name", "host_name"], inplace=True)
```

```python
#Task 3.1 Identifying outliers using histograph
sns.histplot(x=df['price'])
```

```python
#Task 3.1 Identifying outliers using boxplot
sns.boxplot(x=df['minimum_nights'])
```

```python
#Task 3.2 Outlier treatment using 95th percentile rule
for col in ["price", "minimum_nights"]:
    upper = df[col].quantile(0.95)
    df[col] = df[col].clip(upper=upper)
```

```python
#Task 4.1 dropping neighbourhood column
df.drop(columns='neighbourhood',inplace=True)
```

```python
#Task 4.2 One hot encoding
cols = ["neighbourhood_group","room_type"]
for col in cols:
  print(col,df[col].unique())
df = pd.get_dummies(df,columns=cols,dtype=int)
print()
df.info()
```

```python
#Task 5.1 Seperating cleaned data into x and y
x = df.drop(columns=["price"])
y = df["price"]
```

```python
#Task 5.2 Splitting data into 80% training and 20% testing
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
#Task 6.1 Importing and initializing Linear regression model
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
#Task 6.2 Generating price predictions after training
y_pred = model.predict(X_test)
print(y_pred)

results = pd.DataFrame({'Actual': y_test, 'Predicted': y_pred})
print(results)
```

```python
#Task 6.3 Calculating 3 key metrics
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Squared Error:", mse)
print("Mean Absolute Error:", mae)
print("R-squared Score:", r2)
```

```python
#Task 7.1 scatter plot of Actual Prices vs. Predicted Prices
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.5)
plt.plot([y_test.min(), y_test.max()],[y_test.min(), y_test.max()],color="red")
plt.xlabel("Actual Prices ($)")
plt.ylabel("Predicted Prices ($)")
plt.title("Actual vs Predicted Airbnb Prices")

plt.show()
```
